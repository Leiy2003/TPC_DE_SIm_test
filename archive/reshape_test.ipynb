```python
import numpy as np
import torch
import time
```

```python
def CNN_output_vec_(out):
    event_Num = len(out)
    test_Num = event_Num
    # pos = np.zeros((test_Num, 2))
    # train_Num = 0
    print(event_Num)
    layer0 = [18, 17, 16, 15, 14]
    layer1 = [19, 40, 39, 38, 37, 13]
    layer2 = [20, 41, 56, 55, 54, 36, 12]
    layer3 = [21, 42, 57, 4, 5, 6, 53, 35, 11]
    layer4 = [22, 43, 58, 0, 1, 2, 3, 52, 34, 10]
    layer5 = [23, 44, 59, 7, 8, 9, 63, 51, 33]
    layer6 = [24, 45, 60, 61, 62, 50, 32]
    layer7 = [25, 46, 47, 48, 49, 31]
    layer8 = [26, 27, 28, 29, 30]
    layerlist = [layer0, layer1, layer2, layer3, layer4, layer5, layer6, layer7, layer8]

    xy_index = np.empty((64, 2), int)

    # 将对应的通道对应到9*9矩阵的位置
    for i in range(9):
        layeri = layerlist[i]
        id_x = int((10 - len(layeri)) / 2)
        for num in layeri:
            xy_index[num][0] = id_x
            xy_index[num][1] = i
            id_x += 1

    output_test = np.zeros((test_Num, 10, 10))
    # outpos_test = pos[train_Num : train_Num + test_Num]

    t = time.time()
    for i in range(test_Num):
        xj = xy_index[:, 0]
        yj = xy_index[:, 1]
        tmp = np.zeros((10, 10))
        # 归一化
        tmp[yj, xj] = out[i] / (np.sum(out[i][:64]) + 1e-8)
        output_test[i] = tmp
        if (i + 1) % 10000 == 0:
            print(i + 1, "/", test_Num, ",time:", time.time() - t)
            t = time.time()

    print("done of generating test files")

    output_test = torch.FloatTensor(output_test)
    # output_test为N*9*9矩阵，各个通道值按照对应关系填入对应坐标中,outpos_tes为空白的N*2位置矩阵
    return output_test
```

```python
test_response = np.random.normal(loc = 10, scale = 2, size = (5000000,64))
```

```python
t = time.time()
output_test_= CNN_output_vec_(test_response)
print(time.time() - t)
```

```python
def CNN_output_vec(out):
    event_Num = len(out)
    test_Num = event_Num
    # pos = np.zeros((test_Num, 2))
    # train_Num = 0
    print(event_Num)
    layer0 = [18, 17, 16, 15, 14]
    layer1 = [19, 40, 39, 38, 37, 13]
    layer2 = [20, 41, 56, 55, 54, 36, 12]
    layer3 = [21, 42, 57, 4, 5, 6, 53, 35, 11]
    layer4 = [22, 43, 58, 0, 1, 2, 3, 52, 34, 10]
    layer5 = [23, 44, 59, 7, 8, 9, 63, 51, 33]
    layer6 = [24, 45, 60, 61, 62, 50, 32]
    layer7 = [25, 46, 47, 48, 49, 31]
    layer8 = [26, 27, 28, 29, 30]
    layerlist = [layer0, layer1, layer2, layer3, layer4, layer5, layer6, layer7, layer8]

    # xy_index = np.empty((64, 2), int)
    index = np.empty(64, int)

    # 将对应的通道对应到10*10矩阵的位置
    for i in range(9):
        layeri = layerlist[i]
        id_x = int((10 - len(layeri)) / 2)
        for num in layeri:
            index[num] = id_x + i * 10
            id_x += 1

    channel_eventID = np.arange(event_Num)[:, np.newaxis]

    output_test = np.zeros((event_Num, 100))
    output_test[channel_eventID, index] = test_response
    output_test = output_test.reshape(test_response.shape[0],10,10)
    output_test = output_test / np.sum(output_test,axis = (1,2)).reshape((event_Num, 1, 1)) 
    output_test = torch.FloatTensor(output_test)
    # output_test为N*9*9矩阵，各个通道值按照对应关系填入对应坐标中,outpos_tes为空白的N*2位置矩阵
    return output_test
```
